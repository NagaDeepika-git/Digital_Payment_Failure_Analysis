### 2. Missing Values Analysis
Notice that Failure_Reason has nulls, while all other columns have 0 missing values.

```
missing_counts = raw_df.isnull().sum()
missing_pct = (missing_counts / len(raw_df)) * 100
pd.DataFrame({'missing_count': missing_counts, 'missing_percent': missing_pct.round(2)})
```

### 3. Primary Key Uniqueness & Duplicates
Verify that Transaction_id is unique across all records.

```
print(f'Exact duplicate rows: {raw_df.duplicated().sum()}')
print(f'Unique Transaction IDs: {raw_df["Transaction_id"].nunique()}')
assert raw_df['Transaction_id'].nunique() == len(raw_df)
```
