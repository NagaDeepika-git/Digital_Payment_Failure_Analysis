# 02: Data Cleaning & Preprocessing Pipeline

**Project:** Digital Payment Failure Analysis  
**Objective:** Implement and verify reproducible data transformations: column name standardization, safe datetime casting, semantic imputation for successful records, and automated validation assertions.

```
import pandas as pd
import numpy as np
from src.data_loader import load_raw_data
from src.cleaning import clean_payment_data
from src.validation import run_all_validations

raw_df = load_raw_data()
cleaned_df, audit_log = clean_payment_data(raw_df, output_path=None)
print(f'Initial: {audit_log["initial_rows"]} rows -> Final: {audit_log["final_rows"]} rows')
for t in audit_log['transformations']:
    print(f' - {t["step"]}: {t["detail"]}')
```

### 1. Verify Standardized Headers & Types
Ensure all column names are lower snake_case and timestamps are datetime64[ns].

```
print('Cleaned columns:', cleaned_df.columns.tolist())
print('\nTimestamp dtype:', cleaned_df['transaction_timestamp'].dtype)
print(cleaned_df.dtypes)
```

### 2. Verify Semantic Imputation on Failure Reason
Confirm that SUCCESS rows now carry explicit string 'None' instead of NaN.

```
print(cleaned_df.groupby('transaction_status')['failure_reason'].value_counts())
assert (cleaned_df[cleaned_df['transaction_status'] == 'SUCCESS']['failure_reason'] == 'None').all()
```
